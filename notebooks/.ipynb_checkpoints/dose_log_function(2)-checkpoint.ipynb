```python
from itertools import product  # Cartesian product용

import numpy as np

# 딕셔너리 함수에 시그모이드, 델타, 1함수, 1개만 1, 나머지는 0인 함수를 집어 넣기?

# 1. 시그모이드 함수


def doselog(logx, b, c, d, ec50):
    logx = np.asarray(logx, dtype="float")
    z = b * (logx - ec50)
    z = np.clip(z, -700.0, 700.0)
    return c + (d - c) / (1 + np.exp(z))


def sigmoid_dictionary(logx, b_grid, ec50_grid, c=0.0, d=1.0, normalize=True):

    atoms = []
    for (
        b,
        ec50,
    ) in product(
        b_grid, ec50_grid
    ):  # Cartesian product으로 모든 (b, ec50) 조합
        v = doselog(logx, b=b, c=c, d=d, ec50=ec50).astype(float)
        if normalize:
            n = np.linalg.norm(v)
            if n > 0:
                v = v / n
        atoms.append(v)
    return (
        np.vstack(atoms) if atoms else np.zeros((0, len(logx)))
    )  # 조합이 0개면 (0,len(logx))크기에 텅 빈 행렬..


# 입력한 logx 축 위에서, 파라미터 (슬로프b, 중간점?ec50을 그리드로 훑으면서 시그모이드 곡선을 계산하도록..
# 각 곡선을 1차원 벡터.. l2정규화로 길이가 1이 되게 해서 리스트에 쌓아 행렬로 vstack해서 만듬
# (len(b_grid)*len(ec50_grid), len(logx)) <- 행은 파라미터 조합(atom), 열은 logx의 샘플 지점

# 2. delta함수


def delta_dictionary(logx, delta=None, idxs=None, tie="nearest"):
    x = np.asarray(logx, dtype=float)  # x는 x축 좌표 (ex [-2,-1,0,1,2.2])
    n = x.size  # n은 데이터 개수
    if idxs is None:
        if delta is None:
            raise ValueError(
                "delta_dictionary: delta or idxs must be required."
            )  # nearst-> tau값에 가장 가꾼운 x 선택
        delta = np.atleast_1d(np.asarray(delta, dtype=float))
        tie = str(tie).lower()
        if tie == "nearest":
            idxs = np.abs(x[:, None] - delta[None, :]).argmin(
                axis=0
            )  # left-> delta가 들어갈 위치의 왼쪽 인덱스 선택
        elif tie == "left":
            idxs = np.clip(
                np.searchsorted(x, delta, side="left"), 0, n - 1
            )  # right-> delta가 들어가 위치에 오른쪽 인덱스 선택
        elif tie == "right":
            idxs = np.clip(
                np.searchsorted(x, delta, side="right") - 1, 0, n - 1
            )  # tie로 delta가 x벡터 안에 정확히 없으면 어느쪽으로 선택할지 정하게 함
        else:
            raise ValueError("tie must be 'nearest'|'left'|'right'")
    idxs = np.atleast_1d(idxs).astype(int)
    D = np.zeros((len(idxs), n), dtype=float)
    D[np.arange(len(idxs)), idxs] = 1.0
    return D  # idxs 길이만큼 행을 만들고 각 행은 logx길이와 동일하게 만듦, 또한 지정된 인덱스에만 1을 넣음 #(len(idxs), n)행렬 생


def constant_dictionary(logx, normalize=True):
    v = np.ones_like(logx, dtype=float)  # logx 길이를 가진 array생성후 값을 1로 채
    if normalize:
        v = v / np.linalg.norm(v)  # l2놈 정규화로 벡터길이 1로 만듦
    return v[
        None:
    ]  # (1, len(logx)) 축을 추가해서 딕셔너리 행렬에 붙일 수 있게 행렬로 만듦


def build_dictionary(
    logx,
    atoms=("sigmoid", "delta", "constant"),
    *,
    b_grid=None,
    ec50_grid=None,
    c=0.0,
    d=1.0,
    delta_delta=None,
    delta_idxs=None,
    delta_tie="nearest",
    normalize=True,
    return_metadata=True,
):
    # atoms 중에서 골라서 dictionary matrix를 구성하도록 만듦, 3개중에서 튜플 혹은 리스트로 선택함, atoms를 나중에 늘릴 수 있으니 고려..
    x = np.asarray(logx, dtype=float).ravel()
    nfeat = x.size
    parts, meta = [], []

    if "sigmoid" in atoms:
        D_sig = sigmoid_dictionary(x, b_grid, ec50_grid, c=c, d=d, normalize=normalize)
        parts.append(D_sig)
        for b, e50 in product(b_grid, ec50_grid):
            meta.append(
                {
                    "type": "sigmoid",
                    "b": float(b),
                    "ec50": float(e50),
                    "c": float(c),
                    "d": float(d),
                }
            )

    if "delta" in atoms:
        tie = str(delta_tie).lower() if isinstance(delta_tie, str) else delta_tie
        D_del = delta_dictionary(x, delta=delta_delta, idxs=delta_idxs, tie=tie)
        parts.append(D_del)

        if delta_idxs is not None:
            for k in np.atleast_1d(delta_idxs).astype(int):
                meta.append({"type": "delta", "index": int(k)})
        elif delta_delta is not None:
            for t in np.atleast_1d(delta_delta).astype(float):
                meta.append({"type": "delta", "delta": float(t)})

    if "constant" in atoms:
        D_con = constant_dictionary(x, normalize=normalize)
        parts.append(D_con)
        meta.append({"type": "constant"})

    D = np.vstack(parts) if parts else np.zeros((0, nfeat), dtype=float)
    return (D, meta) if return_metadata else D
```

```python
logx = [-2, -1, 0, 1, 2.2]

D, meta = build_dictionary(
    logx,
    atoms=["sigmoid", "delta", "constant"],
    b_grid=[0.5, 1.0, 4.0],
    ec50_grid=[-1.0, 0.0, 1.0],
    delta_delta=[-1.2, 1.7],  # 아니면 delta_idx=[0,49] 등으로
    delta_tie="nearest",
    normalize=True,
    return_metadata=True,
)

print(D.shape)
print(meta[:5])
```

```python
def dose_dictionary_plot(
    logx, b_grid, ec50_grid, c=0.0, d=1.0, title="Dictionary matrix"
):
    x = np.asarray(logx, dtype=float)

    plt.figure(figsize=(12, 3.4))
    for b, e50 in product(b_grid, ec50_grid):
        basis = doselog(x, b=b, c=c, d=d, ec50=e50)
        plt.plot(x, basis, lw=1.2, alpha=0.9, label=f"b={b}, EC50={e50:.2g}")

    plt.xticks([-1, 0, 1, 2.2], [-1, 0, 1, 2.2])
    plt.xlim(x.min(), x.max())
    plt.ylim(-0.02, 1.02)
    plt.ylabel("Basis value")
    plt.title(title)
    plt.legend(
        ncol=3,
        fontsize=8,
        frameon=False,
        columnspacing=0.8,
        handlelength=1.2,
        handletextpad=0.4,
        borderaxespad=0.2,
        labelspacing=0.25,
        loc="upper left",
        bbox_to_anchor=(0.01, 0.99),
    )
    plt.tight_layout()
    plt.show()
```

```python
logx = np.linspace(-2, 2.2, 200)
b_grid = [0.5, 1.0, 2.0]
ec50_grid = [0.5, 1.0]

D, meta = build_dictionary(
    logx,
    atoms=("sigmoid", "delta", "constant"),
    b_grid=b_grid,
    ec50_grid=ec50_grid,
    delta_delta=[1.2],
    normalize=False,
)

plot_dictionary_matrix(D, logx, meta)
```

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import SparseCoder

np.random.seed(1234)

# log-dose 축
logx = np.linspace(-2, 2.2, 200)
dose = np.exp(logx)

# 시뮬레이션 데이터
y_true = doselog(logx, b=2.2, c=0.0, d=1.0, ec50=0.9)
noise = np.random.normal(0, 0.1, size=logx.size)
y_obs = y_true + noise

b_grid = [0.5, 1.0, 2.0, 4.0]
ec50_grid = [0.0, 0.5, 1.0]

D, meta = build_dictionary(
    logx,
    atoms=("sigmoid", "delta", "constant"),
    b_grid=b_grid,
    ec50_grid=ec50_grid,
    delta_delta=[-1.2],
    normalize=True,
    return_metadata=True,
)
# 1/2||wD-y||^2+alpha*||w||_1 를 푸는 해석기 (여기서 w=coef)
coder = SparseCoder(
    dictionary=D,
    transform_algorithm="lasso_lars",  # 또는 lasso_cd, omp도 사용가능
    transform_alpha=0.05,  # 규제 세기 (크면 더 스파스)
)

coef = coder.transform(y_obs.reshape(1, -1))  # 크기는(1, n_atoms)
coef = coef.ravel()

# 복원
y_hat = (coef @ D).ravel()

# 결과 확인
# 어떤 원자가 선택되었는지 상위 기여도 확인
idx_nz = np.flatnonzero(np.abs(coef) > 1e-6)
top = sorted(idx_nz, key=lambda i: abs(coef[i]), reverse=True)[:10]
print("number of using atoms:", idx_nz.size)
for i in top:
    print(f"[{i:3d}] coef={coef[i]: .4f}  meta={meta[i]}")

# 그래프
plt.figure(figsize=(10, 4))
plt.plot(dose, y_obs, "k.", ms=4, label="Observed")
plt.plot(dose, y_hat, "r-", lw=2, label="Sparse recon")
plt.plot(dose, y_true, "g--", lw=1.5, alpha=0.8, label="Ground truth")
plt.xscale("log")
plt.xlabel("Dose")
plt.ylabel("Response")
plt.ylim(-0.05, 1.05)
plt.legend()
plt.title("Sparse coding with dictionary")
plt.show()
```
